```python
# If elif else statement for marks of the students
# Above 95 - grade A, 85-94 - grade C, 75-84 - grade D, below 75 - grade F
marks = int(input("Enter your marks: "))
if marks > 95:
    print("Grade A")
elif marks >= 85:
    print("Grade C")
elif marks >= 75:
    print("Grade D")
else:
    print("Grade F")
```
