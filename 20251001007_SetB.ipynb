```python
#Q.3 A
#Explain the combination formula C(n,r) = n!/(r!(n-r)!) to a second-year chemical engineering student. Explain why factorial appears in the formula. Do not write Python code.
from math import factorial
n = int(input("Enter the value of n: "))
r = int(input("Enter the value of r: "))

def function_combination(n, r):

    if n < 0 or r < 0:
        return "Error: Inputs must be non-negative integers."
    elif r > n:
        return "Error: invalid input: r cannot be greater than n."
    elif r == 0 or r == n:
        return 1
    else:
        numerator = factorial(n)
        denominator = factorial(r) * factorial(n - r)
        return numerator // denominator
        
       
```
